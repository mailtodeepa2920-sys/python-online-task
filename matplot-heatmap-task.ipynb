```python
#HEAT MAP
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Generate some sample 2D data (5x5 matrix)
data = np.random.rand(5, 5)

# 2. Define labels for rows and columns
x_labels = ['A', 'B', 'C', 'D', 'E']
y_labels = ['V', 'W', 'X', 'Y', 'Z']

# 3. Create the plot
fig, ax = plt.subplots()
im = ax.imshow(data, cmap='viridis')  # 'viridis', 'hot', 'coolwarm', etc.

# 4. Add the labels to the axes
ax.set_xticks(np.arange(len(x_labels)), labels=x_labels)
ax.set_yticks(np.arange(len(y_labels)), labels=y_labels)

# 5. Add a colorbar legend
fig.colorbar(im, ax=ax, label='Intensity Value')

# 6. Add title and display
ax.set_title("Simple Matplotlib Heatmap")
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Generate data
data = np.random.rand(5, 5)
x_labels = ['A', 'B', 'C', 'D', 'E']
y_labels = ['V', 'W', 'X', 'Y', 'Z']

# Create heatmap with labels and cell values (annot=True)
sns.heatmap(data, xticklabels=x_labels, yticklabels=y_labels, annot=True, cmap='Blues')

plt.title("Simple Seaborn Heatmap")
plt.show()
```
