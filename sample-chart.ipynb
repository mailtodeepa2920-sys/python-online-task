```python

import matplotlib.pyplot as plt
import mysql.connector 
import pandas as pd
con= mysql.connector.connect(
        host="localhost",
        user="root",
        password="root",
        database="banking"
    )
query3 = "SELECT AccountType, AVG(Balance) as AvgBalance FROM Accounts GROUP BY AccountType"
df3 = pd.read_sql(query3, con)
plt.figure(figsize=(10, 4.5))

# 1. Plot the stems, markers, and baseline cleanly
markerline, stemlines, baseline = plt.stem(
    df3['TxDate'], df3['DailyVolume'], 
    linefmt='#b2dfdb',   # Soft teal stems
    markerfmt='o',       # Circular caps
    basefmt=' '          # Hidden baseline to keep it clean
)

# 2. Style the visual elements
plt.setp(markerline, markerfacecolor='#00897b', markeredgecolor='#004d40', markersize=8)
plt.setp(stemlines, linewidth=1.5)

plt.title('Systemic Liquidity Velocity Timeline', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Date')
plt.ylabel('Aggregated Daily Throughput (₹)')
plt.xticks(rotation=25)

# Format the vertical axis numbers with clean currency commas
ax = plt.gca()
ax.get_yaxis().set_major_formatter(plt.FuncFormatter(lambda x, loc: f"₹{int(x):,}"))

plt.grid(visible=True, axis='y', linestyle=':', alpha=0.5, color='#cccccc')
plt.tight_layout()
plt.show()
```
